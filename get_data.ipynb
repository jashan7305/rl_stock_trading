```python
import yfinance as yf
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path("data")

STOCKS = {
    "hdfc": "HDFCBANK.NS",
    "infosys": "INFY.NS",
}

START_DATE = "2010-01-01"
END_DATE = None
```

```python
for folder in STOCKS:
    (DATA_DIR / folder).mkdir(parents=True, exist_ok=True)
```

```python
for name, ticker in STOCKS.items():
    print(f"Downloading {ticker}...")

    df = yf.download(
        ticker,
        start=START_DATE,
        end=END_DATE,
        interval="1d",
        auto_adjust=False,
        progress=False,
        multi_level_index=False,
    )

    df = df.reset_index()

    output_path = DATA_DIR / name / f"{ticker.replace('.NS', '')}.csv"
    df.to_csv(output_path, index=False)

    print(f"Saved {len(df):,} rows → {output_path}")
```

```python
hdfc = pd.read_csv(DATA_DIR / "hdfc" / "HDFCBANK.csv")
infosys = pd.read_csv(DATA_DIR / "infosys" / "INFY.csv")

print("HDFC Bank:")
display(hdfc.head())

print("\nInfosys:")
display(infosys.head())
```

```python
print(f"HDFC Bank : {len(hdfc):,} rows")
print(f"Infosys   : {len(infosys):,} rows")
```

```python
for name, df in [("HDFC Bank", hdfc), ("Infosys", infosys)]:
    df["Date"] = pd.to_datetime(df["Date"])

    print(
        f"{name}: "
        f"{df['Date'].min().date()} → {df['Date'].max().date()}"
    )
```

```python
print("HDFC Bank missing values:")
display(hdfc.isnull().sum())

print("\nInfosys missing values:")
display(infosys.isnull().sum())
```
